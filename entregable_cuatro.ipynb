# Crear Conjuntos

```python

def crear_conjuntos_de_peliculas():
    csv_path = '/workspaces/AD20252_1/SRC/clases_ad/static/csv/mpst_full_data.csv'

    df = pd.read_csv(csv_path)

    if 'title' not in df.columns or 'tags' not in df.columns:
        raise ValueError("El CSV debe tener columnas 'title' y 'tags'.")
    
    def parse_tags(tag_string):
        return set(tag.strip() for tag in tag_string.split(',') if tag.strip())
    
    df["tag_set"] = df["tags"].apply(parse_tags)

    group_a = set(df[df["tag_set"].apply(lambda tags: 'romance' in tags)]["title"])
    group_b = set(df[df["tag_set"].apply(lambda tags: 'thriller' in tags)]["title"])
    group_c = set(df[df["tag_set"].apply(lambda tags: 'drama' in tags)]["title"])

    all_movies = group_a.union(group_b).union(group_c)
    a_and_b = group_a.intersection(group_b)
    a_and_c = group_a.intersection(group_c)
    b_and_c = group_b.intersection(group_c)
    common_all = group_a.intersection(group_b).intersection(group_c)

    print("Películas con romance:", group_a)
    print("Películas con thriller:", group_b)
    print("Películas con drama:", group_c)
    print("Todas las películas:", all_movies)
    print("Romance y Thriller:", a_and_b)
    print("Romance y Drama:", a_and_c)
    print("Thriller y Drama:", b_and_c)
    print("En los tres grupos:", common_all)

    print("Ejemplos de películas comunes en A ∩ B ∩ C:")
    print(list(common_all)[:5])

crear_conjuntos_de_peliculas()
```

# Grupo

```python
def analizar_peliculas_por_tags():
    csv_path = '/workspaces/AD20252_1/SRC/clases_ad/static/csv/mpst_full_data.csv'

    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print("Error: No se encontró el archivo.")
        return
    except Exception as e:
        print("Error al leer el archivo:", e)
        return
    
    if 'title' not in df.columns or 'tags' not in df.columns:
        print("Error: Las columnas 'title' o 'tags' no están presentes.")
        return
    
    tag_to_movies = {}

    for _, row in df.iterrows():
        movie = row['title']
        tags = row['tags'].split(',')
        for tag in tags:
            tag = tag.strip()
            if tag not in tag_to_movies:
                tag_to_movies[tag] = set()
            tag_to_movies[tag].add(movie)

    all_movies = set()
    for movie_set in tag_to_movies.values():
        all_movies.update(movie_set)

    print(f"Mostrando los primeros 10 títulos con sus tags asociados:\n")
    for movie in sorted(list(all_movies))[:10]:  # Solo muestra 10
        tags_in = []
        for tag, movie_set in tag_to_movies.items():
            if movie in movie_set:
                tags_in.append(tag)
        tags_str = ', '.join(sorted(tags_in))
        plural = 's' if len(tags_in) > 1 else ''
        print(f'"{movie}" pertenece a tag{plural}: {tags_str}')

    
analizar_peliculas_por_tags()                
```

# Métodos disponibles sobre set

```python
csv_path = '/workspaces/AD20252_1/SRC/clases_ad/static/csv/mpst_full_data.csv'
df = pd.read_csv(csv_path)

peliculas = set(df['title'])
tags_unicos = set(tag.strip() for tag_list in df['tags'] for tag in tag_list.split(','))

peliculas_train = set(df[df['split'] == 'train']['title'])
peliculas_val = set(df[df['split'] == 'val']['title'])

# 1. add
def ejemplo_add():
    tags = tags_unicos.copy()
    tags.add('nuevo_tag')
    return tags

# 2. clear
def ejemplo_clear():
    t = tags_unicos.copy()
    t.clear()
    return t  

# 3. copy
def ejemplo_copy():
    copia = peliculas.copy()
    return copia

# 4. difference
def ejemplo_difference():
    return peliculas_train.difference(peliculas_val)

# 5. difference_update
def ejemplo_difference_update():
    pt = peliculas_train.copy()
    pt.difference_update(peliculas_val)
    return pt

# 6. discard
def ejemplo_discard():
    t = tags_unicos.copy()
    t.discard('romantic')  
    return t

# 7. intersection
def ejemplo_intersection():
    return peliculas_train.intersection(peliculas_val)

# 8. intersection_update
def ejemplo_intersection_update():
    pt = peliculas_train.copy()
    pt.intersection_update(peliculas_val)
    return pt

# 9. isdisjoint
def ejemplo_isdisjoint():
    return peliculas_train.isdisjoint(peliculas_val)

# 10. issubset
def ejemplo_issubset():
    return peliculas_val.issubset(peliculas)

# 11. issuperset
def ejemplo_issuperset():
    return peliculas.issuperset(peliculas_train)

# 12. pop
def ejemplo_pop():
    t = tags_unicos.copy()
    return t.pop(), len(t)  

# 13. remove
def ejemplo_remove():
    t = tags_unicos.copy()
    t.remove('violence')  
    return t

# 14. symmetric_difference
def ejemplo_symmetric_difference():
    return peliculas_train.symmetric_difference(peliculas_val)

# 15. symmetric_difference_update
def ejemplo_symmetric_difference_update():
    pt = peliculas_train.copy()
    pt.symmetric_difference_update(peliculas_val)
    return pt

# 16. union
def ejemplo_union():
    return peliculas_train.union(peliculas_val)

# 17. update
def ejemplo_update():
    pt = peliculas_train.copy()
    pt.update(peliculas_val)
    return pt


funciones = [ejemplo_add,ejemplo_clear,ejemplo_copy,ejemplo_difference,ejemplo_difference_update,ejemplo_discard,ejemplo_intersection,
    ejemplo_intersection_update,ejemplo_isdisjoint,ejemplo_issubset,ejemplo_issuperset,ejemplo_pop,ejemplo_remove,ejemplo_symmetric_difference,
    ejemplo_symmetric_difference_update,ejemplo_union,ejemplo_update]

for f in funciones:
    try:
        print(f"\n{f.__name__}:\n", f())
    except Exception as e:
        print(f"\n{f.__name__} - Error:", e)
```

# Operaciones de álgebra de conjuntos

```python
def algebra_conjuntos_peliculas():

    csv_path = '/workspaces/AD20252_1/SRC/clases_ad/static/csv/mpst_full_data.csv'
    df = pd.read_csv(csv_path)

    tag_romantic = set(df[df['tags'].str.contains('romantic', na=False)]['title'])
    tag_drama = set(df[df['tags'].str.contains('dramatic', na=False)]['title'])

    print("Cantidad de películas con tag 'romantic':", len(tag_romantic))
    print("Cantidad de películas con tag 'dramatic':", len(tag_drama))

    # Álgebra de conjuntos
    
    print("\nUnión (romantic ∪ dramatic):")
    print(tag_romantic.union(tag_drama))

    print("\nDiferencia (romantic - dramatic):")
    print(tag_romantic.difference(tag_drama))

    print("\nIntersección (romantic ∩ dramatic):")
    print(tag_romantic.intersection(tag_drama))

    print("\n¿Disjuntos? romantic.isdisjoint(dramatic):")
    print(tag_romantic.isdisjoint(tag_drama))

    print("\nromantic.issubset(todos):")
    print(tag_romantic.issubset(set(df['title'])))

    print("\nromantic.issuperset(intersection):")
    print(tag_romantic.issuperset(tag_romantic.intersection(tag_drama)))

    print("\nDiferencia simétrica (romantic △ dramatic):")
    print(tag_romantic.symmetric_difference(tag_drama))

algebra_conjuntos_peliculas()
```

# Operaciones con conjuntos

```python
def operaciones_con_conjuntos_peliculas():

    csv_path = '/workspaces/AD20252_1/SRC/clases_ad/static/csv/mpst_full_data.csv'
    df = pd.read_csv(csv_path)

    group_a = set(df[df['tags'].str.contains('romantic', na=False)]['title'])
    group_b = set(df[df['tags'].str.contains('violence', na=False)]['title'])

    print("Inicialmente:")
    print(f"group_a (romantic): {len(group_a)} películas")
    print(f"group_b (violence): {len(group_b)} películas\n")

    # len()
    print(f"len(group_a): {len(group_a)}")

    # add()
    group_a.add('Película Inventada')
    print(f"\ngroup_a.add('Película Inventada')")
    print(f"group_a contiene 'Película Inventada'? {'Película Inventada' in group_a}")

    # remove()
    try:
        group_a.remove('Titanic')  # solo si está
        print("\ngroup_a.remove('Titanic') -> OK")
    except KeyError:
        print("\ngroup_a.remove('Titanic') -> ERROR: 'Titanic' no estaba en el grupo")

    # pop()
    popped = group_a.pop()
    print(f"\ngroup_a.pop() -> '{popped}' eliminado")

    # discard()
    group_a.discard('Película No Existente')
    print("group_a.discard('Película No Existente') -> no error si no existe")

    # Comparación >
    print(f"\n¿group_a > group_b? -> {group_a > group_b}")

    # Operador +
    try:
        combinado = group_a + group_b  # Esto da error
    except TypeError as e:
        print("\ngroup_a + group_b -> ERROR:", e)

    # Operador or
    print("\ngroup_a or group_b ->", group_a or group_b)

    # Operador and
    print("\ngroup_a and group_b ->", group_a and group_b)

    # Mostrar ambos conjuntos finales
    print("\nConjuntos finales:")
    print("group_a:", list(group_a)[:5], "...")
    print("group_b:", list(group_b)[:5], "...")

operaciones_con_conjuntos_peliculas()
```

# Frozenset 

```python
def ejemplo_frozenset_peliculas():
    
    csv_path = '/workspaces/AD20252_1/SRC/clases_ad/static/csv/mpst_full_data.csv'
    df = pd.read_csv(csv_path)

    tags_unicos = set()
    for tag_list in df['tags']:
        for tag in tag_list.split(','):
            tags_unicos.add(tag.strip())

    print(f"Total de tags únicos: {len(tags_unicos)}")

    # Crear un frozenset
    tags_frozen = frozenset(tags_unicos)
    print(f"\nfrozenset creado con tags:\n{tags_frozen}\n")

    # Intentar modificarlo (esto debe fallar)
    try:
        tags_frozen.add('nuevo_tag')
    except AttributeError as e:
        print(f"Error al intentar agregar: {e}")

    try:
        tags_frozen.remove('romantic')
    except AttributeError as e:
        print(f"Error al intentar eliminar: {e}")

    # Verificar existencia de un tag
    print("\n'romantic' in tags_frozen:", 'romantic' in tags_frozen)

    # Crear otro conjunto para operaciones
    tags_extra = frozenset(['romantic', 'comedy', 'horror', 'fiction'])

    # Mostrar operaciones posibles
    print("\nIntersección con otro frozenset:")
    print(tags_frozen.intersection(tags_extra))

    print("\nUnión con otro frozenset:")
    print(tags_frozen.union(tags_extra))

    print("\nDiferencia con otro frozenset:")
    print(tags_frozen.difference(tags_extra))

    print("\nDiferencia simétrica:")
    print(tags_frozen.symmetric_difference(tags_extra))

    print("\nLargo del frozenset:", len(tags_frozen))

    print("\nMétodos disponibles en frozenset:")
    print([m for m in dir(tags_frozen) if not m.startswith('__')])

ejemplo_frozenset_peliculas()    
```
