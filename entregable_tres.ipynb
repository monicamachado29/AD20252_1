# 10_Utiliza idxmax() para encontrar el índice del vino con el precio más alto y luego utiliza loc para obtener toda la información de ese vino específico

```python
ruta_csv = "/workspaces/AD20252_1/SRC/clases_ad/static/csv/winemag-data_first150k.csv"
review = pd.read_csv(ruta_csv)
indice_mas_caro = review['price'].idxmax()
vino_mas_caro = review.loc[indice_mas_caro]
print("Vino con el precio más alto:")
print(vino_mas_caro)
```

# 11_Tipos de uva más comunes en California

```python
ruta_csv = "/workspaces/AD20252_1/SRC/clases_ad/static/csv/winemag-data_first150k.csv"
review = pd.read_csv(ruta_csv)
vinos_california = review[review['province'] == 'California']
tipos_uva_comunes = vinos_california['variety'].value_counts()
print("Tipos de uva más comunes en California:")
print(tipos_uva_comunes)
```
