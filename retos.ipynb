```python
import pandas as pd
import numpy as np
```

# Crear DAataframe


```python
data = {
    'nombre': ['Ana', 'Luis', 'Carlos', 'María', 'Elena'],
    'edad': [25, 35, 42, 29, 50],
    'ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Bilbao']
}

df = pd.DataFrame(data)
```

# Filtar clientes mayores de 30 años 

```python
df_filtrado = df[df['edad'] > 30]
```

# agregar columna rango

```python
def clasificar_edad(edad):
    if edad < 40:
        return 'Joven'
    else:
        return 'Adulto'

df_filtrado['rango'] = df_filtrado['edad'].apply(clasificar_edad)

print(df_filtrado)
```

# RETO 2

```python
data = {
    'producto': ['Laptop', 'Teléfono', 'Tablet', 'Monitor', 'Teclado'],
    'precio': [1200, 800, 400, 300, 100]
}

df = pd.DataFrame(data)
```

# Normaliza precios entre 0 y 1

```python
min_precio = df['precio'].min()
max_precio = df['precio'].max()

df['precio_normalizado'] = df['precio'].apply(lambda x: (x - min_precio) / (max_precio - min_precio))

print(df)
```

# RETO 3 

Generar etiquetas de ingreso

```python
data = {
    'empleado': ['Juan', 'Ana', 'Luis', 'Marta', 'Carlos'],
    'salario': [1500, 2500, 4200, 3900, 1800]
}

df = pd.DataFrame(data)
```

# Nivel ingreso

```python
df['nivel_ingreso'] = df['salario'].apply(
    lambda x: 'alto' if x > 4000 else ('medio' if x >= 2000 else 'bajo')
)

print(df)
```

# Reto 4  - Transformar texto

```python
data = {'producto': ['manzana roja', 'jugo de naranja', 'pan integral', 'leche descremada']}
df = pd.DataFrame(data)
```

# Convertir la primera letra de cada palabra a mayúscula

```python
df['producto'] = df['producto'].apply(lambda x: x.title())
print(df)
```

# Reto 5 - Limpieza de datos nulos

```python
data = {
    'cliente': ['Ana', 'Luis', 'María', 'Carlos'],
    'correo': ['ana@gmail.com', None, 'maria@hotmail.com', np.nan]
}
df = pd.DataFrame(data)
```

# Reemplazar valores nulos

```python
df['correo'] = df['correo'].apply(lambda x: x if pd.notnull(x) else 'desconocido@correo.com')
print(df)
```
