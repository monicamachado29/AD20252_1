```python
class Persona ():
    def __init__(self,nombre,apellido,edad,genero,altura):
        self.nombre=nombre
        self.apellido=apellido
        self.edad=edad
        self.genero=genero
        self.altura=altura
    def caminar (self):
        print (self.nombre + " " + self.apellido + " está caminando para la IUDigital.")
    
  
```

```python
persona1 = Persona("Lucía", "González", 25, "mujer", 1.68)
persona2 = Persona("Carlos", "Ramírez", 16, "hombre", 1.75)
```

```python
persona1.caminar()
persona2.caminar()
```
